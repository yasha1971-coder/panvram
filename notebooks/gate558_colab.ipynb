# panvram 1.0 - the 558-assembly gate on Colab (GPU sm_80+)

The HPRC cohort (558 assemblies, refrel3 v1 **q4k**) resident on one GPU: open time, VRAM, `sample(1024, 8192)` and 1 000
fetches checked against the CPU decoder, and every assembly decoded in full on the GPU with the FASTA XXH3 == the
manifest's source XXH3 (no source FASTA needed).

**Dataset DOI: `10.5281/zenodo.XXXXXXX` (placeholder - set `DATASET_RECORD` below when the record exists).**
The record is expected to hold `cohort558_q4k.tar`, `cohort558_q4k.tar.sha256`, `manifest_v1.tsv` and `panvram-1.0.0.tar.gz`.
Runtime: a GPU runtime with about 30 GB of free disk; the run ends with `DONE — выключи runtime`.

```python
# 1. Drive (for the log) and settings
from google.colab import drive
drive.mount('/content/drive')
DATASET_RECORD = "XXXXXXX"            # Zenodo record id of DOI 10.5281/zenodo.XXXXXXX (placeholder)
import os
os.environ["DATASET_URL"] = f"https://zenodo.org/records/{DATASET_RECORD}/files"
os.environ["DL"] = "/content/panvram_dataset"
```

```python
%%bash
# 2. Download the dataset record (or copy it to MyDrive/panvram/ by hand and skip this cell)
set -euo pipefail
ulimit -c 0
mkdir -p $DL
for f in cohort558_q4k.tar.sha256 manifest_v1.tsv panvram-1.0.0.tar.gz cohort558_q4k.tar; do
  [ -s $DL/$f ] || curl -fL --retry 3 -o $DL/$f "$DATASET_URL/$f?download=1"
done
ls -l $DL
```

```python
%%bash
# 3. The gate: run_colab_cohort558.sh from the package (tar sha256 checked, archives sha256 == manifest, then the
#    GPU measurements and the full decode of all 558); log in MyDrive/aceapex_logs/panvram_cohort558_<date>.txt
set -euo pipefail
ulimit -c 0
mkdir -p /content/pv && tar -xzf $DL/panvram-1.0.0.tar.gz -C /content/pv
PV_TAR=$DL/panvram-1.0.0.tar.gz COHORT_TAR=$DL/cohort558_q4k.tar MANIFEST=$DL/manifest_v1.tsv \
  bash /content/pv/panvram/run_colab_cohort558.sh
```
